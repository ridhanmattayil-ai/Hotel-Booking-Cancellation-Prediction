# Insights

## Insight 1: Lead Time Strongly Influences Cancellations

Bookings made far in advance tend to have a higher cancellation rate compared to bookings made closer to the arrival date. A longer lead time provides customers with more opportunities to change their travel plans, resulting in an increased likelihood of cancellation.

---

## Insight 2: Larger Travel Groups Show Slightly Higher Cancellation Tendencies

The engineered **Total Guests** feature revealed that bookings involving larger groups have a slightly higher cancellation rate. Group travel often requires coordination among multiple individuals, making reservations more susceptible to schedule changes and cancellations.

---

## Insight 3: Length of Stay Affects Booking Behavior

Analysis of the **Total Nights** feature indicates that booking duration influences cancellation patterns. Longer stays generally involve higher travel costs and more extensive planning, which can impact the likelihood of cancellation.

---

## Insight 4: Family and Non-Family Bookings Exhibit Different Behaviors

The **Family Booking Indicator** highlighted differences between family and non-family reservations. Family travel often depends on school schedules, holidays, and group planning, which may contribute to distinct booking and cancellation patterns.

---

## Insight 5: Customer Booking History Provides Valuable Predictive Information

The **Customer History** feature demonstrated that previous booking behavior can help explain future cancellation tendencies. Guests with prior booking experience may exhibit more consistent reservation behavior than first-time customers.

---

## Insight 6: Most Bookings Are Correctly Classified by the Models

Both Logistic Regression and Decision Tree models achieved good overall classification performance. The confusion matrices show that the models successfully identify a large proportion of non-cancelled bookings while also detecting a meaningful number of cancelled reservations.

---

## Insight 7: Decision Tree Outperformed Logistic Regression

Model evaluation results indicate that the Decision Tree achieved slightly higher Accuracy, Recall, and F1-Score than Logistic Regression. The Decision Tree was more effective at identifying cancelled bookings, making it the stronger model for this prediction task.

---

# Limitations

- The dataset may not capture all factors influencing booking cancellations.
- Some variables contained missing values that required assumptions during preprocessing.
- Class imbalance may affect the model's ability to identify cancelled bookings.
- Only two classification algorithms were evaluated.
- External factors such as economic conditions, weather conditions, and travel restrictions were not included in the analysis.

---

# Possible Improvements

- Evaluate additional machine learning models such as Random Forest, XGBoost, and Gradient Boosting.
- Perform hyperparameter tuning to improve model performance.
- Apply advanced techniques to address class imbalance.
- Incorporate external data sources that may influence booking behavior.
- Conduct feature selection to identify the most influential predictors.

---

# Conclusion

The objective of this project was to develop a predictive analytics solution capable of identifying hotel booking cancellations using historical reservation data.

The project followed a complete machine learning workflow, including Exploratory Data Analysis (EDA), data cleaning, preprocessing, feature engineering, predictive modeling, and model evaluation. Several meaningful features were created, including **Total Guests**, **Total Nights**, **Family Booking Indicator**, and **Customer History**, which helped capture customer and reservation behavior more effectively.

Two classification models were implemented: **Logistic Regression** and **Decision Tree**. Both models demonstrated strong predictive performance; however, the Decision Tree achieved the best overall results with higher Accuracy, Recall, and F1-Score. The model was particularly effective at identifying cancelled bookings while maintaining competitive precision.

The analysis revealed that factors such as booking lead time, group size, stay duration, family travel patterns, and customer booking history influence cancellation behavior. These insights can help hotel management anticipate cancellations, optimize room allocation, improve revenue planning, and develop strategies to reduce the impact of lost bookings.

Overall, the **Decision Tree model is recommended as the preferred solution for predicting hotel booking cancellations**, as it delivered the strongest predictive performance among the models evaluated in this project.
