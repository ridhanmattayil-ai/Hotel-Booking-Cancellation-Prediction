# **Data Cleaning & Preprocessing**



The dataset contains missing values, duplicate records, inconsistent values, and categorical variables that require transformation before machine learning can be applied.

The objectives of this stage are:

- Handle missing values
- Remove duplicate records
- Fix inconsistent data
- Convert data types where necessary
- Encode categorical variables
- Prepare the dataset for feature engineering and modeling

## **Load DataSet**

```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv(r"C:\Users\user\OneDrive\Desktop\Machine Learning Project\Data\hotel_bookings.csv")
df.head()
```

## **Checking Missing Values**

```python
df.isnull().sum().sort_values(ascending=False)
```

## **Handle Missing Values**

```python
# children
df['children'] = df['children'].fillna(0)

# country
df['country'] = df['country'].fillna('Unknown')

# agent
df['agent'] = df['agent'].fillna(0)

# company
df.drop('company', axis=1, inplace=True)
```

Agent and company columns contain a large number of missing values.

These missing values do not necessarily indicate missing information but often represent bookings made without an associated travel agent or company.

Therefore, missing values were replaced with 0.

## **Verifying Missing Values**

```python
df.isnull().sum().sort_values(ascending=False)
```

## Missing Value Treatment

The following actions were taken:

- Missing values in `children` were replaced with 0.
- Missing values in `country` were replaced with "Unknown".
- Missing values in `agent` were replaced with 0.
- The `company` column was removed because approximately 94% of its values were missing.

These decisions were made to preserve useful information while reducing noise.

## **Check Duplicate Records**

```python
df.duplicated().sum()
```

## **Removing Duplicates**

```python
df.drop_duplicates(inplace=True)
df.duplicated().sum()
```

## **Outlier Detection And Removing**

```python
outlier_columns = ['lead_time', 'adr']

for col in outlier_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - (1.5 * IQR)
    upper_bound = Q3 + (1.5 * IQR)

    df = df[
        (df[col] >= lower_bound) &
        (df[col] <= upper_bound)
    ]
```

## **Fix Inconsistent Values**

```python
df[
    (df['adults'] == 0) &
    (df['children'] == 0) &
    (df['babies'] == 0)
].shape

```

```python
df = df[
    ~(
        (df['adults'] == 0) &
        (df['children'] == 0) &
        (df['babies'] == 0)
    )
]
```

## **Check Data Types**

## **Convert To Date Column**

```python
df['reservation_status_date'] = pd.to_datetime(
    df['reservation_status_date']
)
df.info()
```

## **Check Cardinality Before Encoding**

```python
categorical_columns = df.select_dtypes(include='string').columns

for col in categorical_columns:
    print(col, ":", df[col].nunique())
```

# **Label Encoding**

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

label_cols = [
    'hotel',
    'deposit_type',
    'customer_type',
    'reservation_status'
]

for col in label_cols:
    df[col] = le.fit_transform(df[col])

df
```

# **One-Hot Encoding**

```python
one_hot_cols = [
    'meal',
    'market_segment',
    'distribution_channel'
]

df = pd.get_dummies(
    df,
    columns=one_hot_cols,
    drop_first=True,
    dtype=int
)
```

```python
print("Shape After Cleaning:", df.shape)

df.head()
```

```python
# Convert month
month_mapping = {
    'January':1,
    'February':2,
    'March':3,
    'April':4,
    'May':5,
    'June':6,
    'July':7,
    'August':8,
    'September':9,
    'October':10,
    'November':11,
    'December':12
}
df['arrival_date_month'] = df['arrival_date_month'].map(month_mapping)

# Drop country
df.drop('country', axis=1, inplace=True)

# One-hot encode room types
df = pd.get_dummies(
    df,
    columns=['reserved_room_type',
             'assigned_room_type'],
    drop_first=True,
    dtype=int
)
```
